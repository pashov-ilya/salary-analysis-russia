### Открываем 2 файла и заносим данные в один большой словарь

```python
with open('my_date_2022.csv', mode='r', encoding ='UTF-8') as infile: # проверяем, открывается ли корректно файл 
    lines = infile.readlines()
    num_lines = len(lines) # подсчитываем количество строк в файле (их 101).
    print(f'количество строк в файле с данными 2022 года: {num_lines}')
    print(infile)
```

### Всего строк в файле 101. Файл за 2022 год открывается корректно.

```python
data=[] # создали список, куда будем добавлять каждый элемент файла
with open('my_date_2022.csv', encoding ='UTF-8') as infile:
    for line in infile:
        values = line.strip().split(',')  # Разделяем строку по запятой и убираем лишние пробелы, пропуски по бокам элементов
        clean_values = []  # Создаем временный список для очищенных значений
        for value in values:
            value = value.replace('"', '') # удаляем двойные кавычки
            clean_values.append(value.strip())  # Очищаем пробелы и добавляем в список(без этого почему-то отсавались лишние пробелы у некоторых элементов)
        data.append(clean_values)  # Добавляем очищенную строку в `data`      
print(*data, sep=', ') # вывели получившийся результат для оценки корректности 
```

```python
dictionary = {data[1][5]: {}} # создали словарь с ключом 2022 года и значением, который является пустой словарь, в который добавим субъекты и их среднюю заработную плату 
rows = data[2:-2]  # строки с регионами

for row in rows:
    subject = row[0]  # Наименование субъекта - первый элемент строки
    value_2022 = row[5]  # последний элемент списка, который имеет среднюю зп. субъекта за 2022 год
    
    if subject and value_2022:  # Проверяем, что наименование субъекта не пустое
        dictionary[data[1][5]][subject] = value_2022  # Добавляем в словарь значение 2022 года

# проверяем результат
print(dictionary)
```

### Проделываем то же самое, чтобы добавить в словарь среднюю заработную плату работников субъектов за 2023 год

```python
with open('my_date_2023.csv', mode='r', encoding ='UTF-8') as file: # прописываем все параметры в явном виде
    lines_2023 = file.readlines()
    num_lines_2023 = len(lines_2023) # подсчитываем количество строк в файле (должно быть 101).
    print(f'количество строк в файле с данными 2023 года: {num_lines_2023}')
    print(file)
```

### В файле с данными за 2023 год строк также 101, файл открывается корректно.

```python
data_2=[]
with open('my_date_2023.csv', encoding='UTF-8') as file:
    for line_2 in file:
        values_2 = line_2.strip().split(',') 
        clean_values_2 = []
        for value_2 in values_2:
            value_2 = value_2.replace('"', '')
            clean_values_2.append(value_2.strip())
        data_2.append(clean_values_2) 
print(data_2, sep=', ')
```

```python
dictionary[data_2[1][1]] = {}  # Добавляем новый ключ ('2023'), основываясь на втором списке из data_2

# Проходим по остальным строкам в data_2 и добавляем их в новый вложенный словарь
for row_2 in data_2[2:-2]:  # строки с регионами
    subject_2 = row_2[0] # Субъект
    number = row_2[1] # средний доход субъекта
    # Добавляем пары "субъект-доход" в новый вложенный словарь для ключа 2023 года
    if subject_2 and number:
        dictionary[data_2[1][1]][subject_2] = number

# Выводим итоговый словарь
print(dictionary)
```

#### Проверяем, полностью ли совпадают данные за 2022 и 2023 года (имеются ли все регионы) для дальнейшего корректного подсчёта статистик

```python
set_2022 = set(dictionary["2022"].keys())  # регионы за 2022 год
set_2023 = set(dictionary["2023"].keys())  # регионы за 2023 год

if set_2022 == set_2023:
    print('Состав регионов РФ полностью совпадает в данных 2022 и 2023 годов.')
else:
    print('В данных 2022 и 2023 годов есть различия в составе регионов РФ.')
```

### Переходим к анализу данных

##### Анализ статистических параметров помогает глубже понять распределение и динамику заработных плат в регионах России за 2022 и 2023 годы.



#### Отбираем необходимые данных двух лет и заносим их в 2 списка по годам, Исключая значения для федеральных округов и Российской Федерации

```python
# создаём список с ненужными нам ключами(фед. округи и РФ)
stopwords = ['Российская Федерация', 'Центральный федеральный округ', 'Северо-Западный федеральный округ', 'Южный федеральный округ', 'Северо-Кавказский федеральный округ', 'Приволжский федеральный округ', 'Уральский федеральный округ', 'Сибирский федеральный округ', 'Дальневосточный федеральный округ']
# Создаем пустые списки для данных
data_2022 = []
data_2023 = []

# Заполняем data_2022, исключая значения из stopwords
for key, value in dictionary['2022'].items():
    if key not in stopwords:  # Проверяем, что ключ не находится в списке исключений
        data_2022.append(int(value))  # Преобразуем значение в число и добавляем в список

# Заполняем data_2023, исключая значения из stopwords
for key, value in dictionary['2023'].items():
    if key not in stopwords:  # Проверяем, что ключ не находится в списке исключений
        data_2023.append(int(value))  # Преобразуем значение в число и добавляем в список

# Выводим полученные списки для проверки
print("Данные 2022 года:", data_2022)
print("Данные 2023 года:", data_2023)
```

### Создаём функцию, которая будет определять род существительного, вычислять разницу значений за 2023 и 2022 года и определять, повысилились или понизились показатели

```python
def compare_values(val_2023,val_2022, kind): # kind отвечает за род существительного
    difference = val_2023 - val_2022 #создаём переменную, которая вычисляет разницу двух лет
    if kind == 'm':
        increase, decrease, same = "увеличился", "снизился", "не изменился"
    elif kind == 'f':
        increase, decrease, same = "увеличилась", "снизилась", "не изменилась"
    else:
        increase, decrease, same = "увеличилось", "снизилось", "не изменилось"
    
    if difference > 0:
        trend = f"{increase} на {round(difference,2)}"
    elif difference < 0:
        trend = f"{decrease} на {round(abs(difference,2))}"
    else:
        trend = same
    return trend       
```

### Среднее значение

```python
mean_2022 = round(sum(data_2022) / len(data_2022), 2) # находим среднее значение за 2022 год и округляем до 2 цифр после запятой
mean_2023 = round(sum(data_2023) / len(data_2023), 2) # находим среднее значение за 2023 год и округляем до 2 цифр после запятой

mean_result = compare_values(mean_2023,mean_2022, kind = 'n') # среднее значение среднего рода

print(f'среднее значение 2022 года: {mean_2022} руб.')
print(f'среднее значение 2023 года: {mean_2023} руб.') # выводим значения 
print(f'среднее значение {mean_result} руб.')
```

#### Показатель отражает общую тенденцию заработных плат по всем регионам. 
Средняя заработная плата по регионам увеличилась на 8329,56 руб (с 56664,77 руб. в 2022 году до 64994,33 руб. в 2023 году). Это могло произойти из-за инфляции, повышения зарплат или экономического роста.

### Медиана

```python
sorted_2022 = sorted(data_2022) # отсортированный список по возрастанию 2022 года
sorted_2023 = sorted(data_2023) # отсартированный список по возрастанию 2023 года

n_2022 = len(sorted_2022) # кол-во значений 2022 года
n_2023 = len(sorted_2023) # кол-во значений 2023 года

if n_2022 % 2 == 0: # проверяем чётное ли кол-во значений для медианы
    median_2022 = (sorted_2022[n_2022 // 2 - 1] + sorted_2022[n_2022 // 2]) / 2 # находим медиану, в квадратных скобках индекс нужного значения
else: # если кол-во значений для медианы нечётно
    median_2022 = sorted_2022[n_2022 // 2] # находим медиану, в квадратных скобках n_2022 // 2 — индекс среднего элемента, который нам нужен 
# проделываем то же самое с 2023 годом
if n_2023 % 2 == 0:
    median_2023 = (sorted_2023[n_2023 // 2 - 1] + sorted_2023[n_2023 // 2]) / 2
else:
    median_2023 = sorted_2023[n_2023 // 2]

median_result = compare_values(median_2023,median_2022, kind = 'f') # медиана женского рода
# выводим ответы
print(f'медиана 2022 года: {median_2022} руб.')
print(f'медиана 2023 года: {median_2023} руб.')
print(f"медиана {median_result} руб.")
```

#### Медиана показывает среднюю заработную плату региона, которая делит выборку на две равные части. Она менее чувствительна к аномально высоким значениям (например, зарплатам в столичных регионах).
Рост медианы на 7756 руб. (с 47780 руб в 2022 году до 55536 руб. в 2023 году) также свидетельствует о том, что заработные платы увеличиваются не только у высокооплачиваемых работников, но и в регионах с более низкими доходами.

### Минимум и максимум

```python
# находим минимуи и максимум средних заработных плат работникрв за 2022 год
min_2022 = min(data_2022)
max_2022 = max(data_2022)
# находим минимуи и максимум средних заработных плат работникрв за 2023 год
min_2023 = min(data_2023)
max_2023 = max(data_2023)

min_result = compare_values(min_2023,min_2022, kind = 'm') # находим разницу минимальной зароботной платы между 2023 и 2022 годом
max_result = compare_values(max_2023,max_2022, kind = 'm') # находим разницу максимальной зароботной платы между 2023 и 2022 годом
# выводим значения
print(f'минимум 2022 года: {min_2022} руб., максимум 2022 года: {max_2022} руб.')
print(f'минимум 2023 года: {min_2023} руб., максимум 2023 года: {max_2023} руб.')
print(f'минимум {min_result} руб.')
print(f'максимум {max_result} руб.')
```

#### Минимум показывает уровень зарплат в наименее экономически развитых регионах, что указывает на потенциальные социально-экономические проблемы в этих субъектах. Максимум же демонстрирует уровень зарплат в экономически развитых регионах, таких как Москва или регионы с высокими зарплатами в ключевых отраслях.
Минимальная заработная плата по регионам увеличилась на 3024 руб. (с 32801 руб. в 2022 году до 35825 руб. в 2023 году).
Максимальная заработная плата увеличилась на 18469 руб. (с 140602 руб. в 2022 году до 159071 руб. в 2023 году).
Увеличение максимума и минимума показывает, что рост зарплат происходит как в отстающих, так и в лидирующих регионах.


### Стандартное отклонение

```python
dispersion_2022 = sum((x - mean_2022) ** 2 for x in data_2022) / len(data_2022) # находим дисперсию 2022 года
dispersion_2023 = sum((x - mean_2023) ** 2 for x in data_2023) / len(data_2023) # находим дисперсию 2023 года
print(f"дисперсия для проверки, что правильно выводит: {dispersion_2022}") 
standard_deviation_2022 = round(dispersion_2022 ** 0.5, 2) # находим стандартное отклонение 2022 года
standard_deviation_2023 = round(dispersion_2023 ** 0.5, 2) # находим стандартное отклонение 2023 года

standard_deviation_result = compare_values(standard_deviation_2023, standard_deviation_2022, kind = 'n')
# выводим значения
print(f'стандартное отклонение 2022 года: {standard_deviation_2022} руб.')
print(f'стандартное отклонение 2023 года: {standard_deviation_2023} руб.')
print(f'стандартное отклонение {standard_deviation_result} руб.')
```

#### Этот показатель измеряет разброс зарплат по регионам относительно среднего значения.
Рост стандартного отклонения на 2247.65 руб. (с 23221.76 руб. в 2022 году до 25469.41 руб. в 2023 году) указывает на увеличение неравенства в доходах между регионами.

### Pазмах значений

```python
range_2022 = max_2022 - min_2022 # находим размах значений за 2022 год
range_2023 = max_2023 - min_2023 # находим размах значений за 2023 год

range_result = compare_values(range_2023, range_2022, kind = 'm')

print(f'размах среднего дохода субъектов в 2022 году: {range_2022} руб.')
print(f'размах среднего дохода субъектов в 2023 году: {range_2023} руб.')
print(f'размах значений {range_result} руб.')
```

#### Размах отражает разницу между самым высоким и самым низким значением.
Его увеличение за год на 15445 руб. (с 107801 руб. в 2022 году до 123246 руб. в 2023 году) указывает на усиливающееся экономическое неравенство, хотя общий рост доходов и наблюдается.

### В совокупности эти показатели помогают проанализировать динамику доходов, выявить ключевые тенденции и разработать рекомендации для выравнивания экономических условий в регионах.

## Cоставляем рейтинг топ-10 субъектов с максимальными средними заработными платами работников по РФ за 2022 и 2023 года

### Рейтинг 2022 года

```python
# Список пар (регион, зп.) для 2022 года
d_2022 = list(dictionary['2022'].items())

# Сортировка списка по убыванию зп.
for i in range(len(d_2022)):
    for j in range(len(d_2022) - i - 1): # - i - 1 уменьшает количество элементов, которые нужно проверять, потому что на конце списка уже будет расположен наименьший элемент.
        if int(d_2022[j][1]) < int(d_2022[j + 1][1]):
            d_2022[j], d_2022[j + 1] = d_2022[j + 1], d_2022[j] # меняем элементы местами, чтобы большая заработная плата была перед меньшей

# Отбираем топ-10 регионов с самыми высокими зп. за 2022 год
top_2022 = d_2022[:10]
print(*top_2022, sep='\n') # выводим рейтинг 2022 года
```

### рейтинг 2023 года 

```python
#(проделываем то же самое)
# Список пар (регион, зп.) для 2023 года
d_2023 = list(dictionary['2023'].items())

# Сортировка списка data_2023 по убыванию заработной платы
for i in range(len(d_2023)):
    for j in range(len(d_2023) - i - 1):
        if int(d_2023[j][1]) < int(d_2023[j + 1][1]):
            d_2023[j], d_2023[j + 1] = d_2023[j + 1], d_2023[j]

# Выбор топ-10 регионов с самыми высокими зп. за 2023 год
top_2023 = d_2023[:10]
print(*top_2023, sep='\n')
```
